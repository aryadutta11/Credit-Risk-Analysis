```python
import pandas as pd
import numpy as np
```

```python
df =pd.read_csv(r"C:\Users\User\Desktop\Credit_risk_analysis\data\processed\credit_risk_dataset.csv")
```

```python
#income to loan ratio
df["income_loan_ratio"] = (df["person_income"] / df["loan_amnt"])
```

```python
#monthly income
df["monthly_income"] = df["person_income"] / 12
```

```python
#estimated monthly loan burden
df["loan_income_ratio"] = (df["loan_amnt"] / df["person_income"]) 
```

```python
#income risk category
df["income_risk"] = pd.cut(df["person_income"], bins=[0,30000,60000,100000,float("inf")],labels=["Low Income","Lower-Middle","Upper-Middle","High Income"])
```

```python
df["income_risk"].value_counts()
```

```python
#Loan size category
df["loan_size_category"] = pd.cut(df["loan_amnt"], bins=[0,5000,10000,20000,float("inf")],labels=["Small","Medium","Large","Very Large"])
```

```python
df["loan_size_category"].value_counts()
```

```python
#Age Group
df["age_group"] = pd.cut(
    df["person_age"],
    bins=[17, 25, 35, 50, 65, 100],
    labels=[
        "18-25",
        "26-35",
        "36-50",
        "51-65",
        "66+"
    ]
)
```

```python
df["age_group"].value_counts()
```

```python
#employment category
df["employment_category"] = pd.cut(
    df["person_emp_length"],
    bins=[-1, 1, 5, 10, float("inf")],
    labels=[
        "New Employee",
        "Experienced",
        "Very Experienced",
        "Long Tenure"
    ]
)
```

```python
#previous default flag
df["previous_default"] = np.where(
    df["cb_person_default_on_file"] == "Y",
    1,
    0
)
```

```python
#high dept burden flag
df["high_debt_burden"] = np.where(
    df["loan_percent_income"] >= 0.30,
    1,
    0
)
```

```python
#high interest rate flag
df["high_interest_rate"] = np.where(
    df["loan_int_rate"] >= 15,
    1,
    0
)
```

```python
df.to_csv(
    r"C:\Users\User\Desktop\Credit_risk_analysis\data\processed\credit_risk_features.csv",
    index=False
)

print("Feature engineering completed.")
```
